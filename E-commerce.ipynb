```python
df["revenue"] = df["quantity"] * df["price"]
```

```python
top_products = df.groupby('product_name')['quantity'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_products.plot(kind='barh')

plt.title("Top 10 Products by Demand")
plt.xlabel("Total Quantity Sold")   # ✅ FIX
plt.ylabel("Product Name")

plt.show()
```

```python
payment_sales = df.groupby('payment_method')['revenue'].sum().sort_values()

payment_sales.plot(kind='barh', figsize=(8,5), title="Revenue by Payment Method")

plt.xlabel("Total Revenue")
plt.ylabel("Payment Method")
plt.show() #Revenue by Payment Method
```

```python
category_sales = df.groupby('category')['revenue'].sum().sort_values() 

category_sales.plot(kind='barh', figsize=(10,6), title="Revenue by Category")

plt.xlabel("Revenue")
plt.show() #Revenue by Category
```

```python
brand_sales = df.groupby('brand')['revenue'].sum().sort_values(ascending=False).head(10)

brand_sales.plot(kind='barh', figsize=(10,6), title="Top Brands by Revenue")

plt.xlabel("Revenue")
plt.show() #Brand Performance
```

```python
discount_sales = df.groupby('discount_pct')['revenue'].mean().reset_index()

plt.figure(figsize=(8,5))
sns.lineplot(x='discount_pct', y='revenue', data=discount_sales)

plt.title("Discount vs Average Revenue")
plt.show() #Discount Impact
```

```python
rating_sales = df.groupby('avg_rating')['quantity'].sum().reset_index()

plt.figure(figsize=(8,5))
sns.lineplot(x='avg_rating', y='quantity', data=rating_sales)

plt.title("Rating vs Demand")
plt.show() #Rating vs Sales
```

```python
df.isnull().sum()
```

```python
category_revenue = df.groupby('category')['revenue'].sum()

plt.figure(figsize=(10,8))
plt.pie(category_revenue, labels=category_revenue.index, autopct='%1.1f%%')

plt.title("Revenue Share by Category")

plt.show()
```

```python
top_brands = df.groupby('brand')['revenue'].sum().sort_values(ascending=False).head(5)

plt.figure(figsize=(7,7))
plt.pie(top_brands, labels=top_brands.index, autopct='%1.1f%%')

plt.title("Top 5 Brands Revenue Share")
plt.show() #Top 5 Brands
```
